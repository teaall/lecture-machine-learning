## Exercise 2: Broadcasting 

Broadcasting is a powerful capability of numpy, that, within certain rules, allows operations (such as addition) on two arrays whose shapes do not match.
Numpy enables this by repeating values along dimensions of length 1, and appending such dimensions if necessary.
Broadcasting enables efficient and concise operations on arrays without Python loops and without explicitly inserting or repeating dimensions.
If you are not familiar with broadcasting, we strongly recommend to read through out the very well written numpy tutorial on it (https://numpy.org/devdocs/user/basics.broadcasting.html).

In the following, we will provide a short recap of broadcasting.

First, print things out again so we know what we are talking about 

```python
##### DO NOT CHANGE #####
print("Matrix A: (Shape {})".format(A.shape))
print(A)
print("Vector v: (Shape {})".format(v.shape))
print(v)
print("Vector w: (Shape {})".format(w.shape))
print(w)

##### DO NOT CHANGE #####
```

Let's try something weird and and subtract a vector from a matrix

```python
##### DO NOT CHANGE #####
print(A - v)

##### DO NOT CHANGE #####
```

As we see, numpy did not give us an error but did something - it subtracted the vector from each row of the matrix. This mechanism is called **broadcasting** and it will be your new best friend (you will sometimes also hate it)! 

In general, if the shapes of two numpy arrays do not match, but the non-matching dimension is "1" for either of the arrays the values along that dimension will be broadcasted. Here, numpy implicitly added a axis of size 1 to the vector "v" and broadcasted the values of "v" to all rows of A. This implicity broadcasting can be a bit counter intuitive and is a common source of errors.

Yet, broadcasting is a very important tool to efficiently combine data in the ways you need to and you will see many applications throughout the exercises. As a general tip: **never** use "for-loops" to manipulate your arrays but try to find a way using broadcasting - it is just much much faster. We might deduce points in the exercises for avoidable for-loops.

So let's do the broadcasting more explicitly: 

```python
##### DO NOT CHANGE #####
v1 = v[np.newaxis, :] # equivalent to v[None, :] or np.reshape(v, (1, 3))
print("v1:", v1)
print("Shape:", v1.shape)
print("Subtraction: \n", A - v1)

##### DO NOT CHANGE #####
```

If we want to subtract the vector from each column instead of each row, we need to broadcast along the other axis! 

```python
##### DO NOT CHANGE #####
v2 = v[:, np.newaxis] # equivalent to v[:, None] or np.reshape(v, (3, 1))
print("v2:\n", v2)
print("Shape:", v2.shape)
print("Subtraction: \n", A - v2)

##### DO NOT CHANGE #####
```
